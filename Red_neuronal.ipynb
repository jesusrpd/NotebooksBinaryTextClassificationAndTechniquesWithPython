```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# Cargar el dataset MNIST
# Dividimos el dataset, x_train(datos entrenamiento), y_train (etiquetas), x_test e y_test (para testear el modelo)
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalizar los datos (de 0-255 a 0-1), asi queda en una escala más fácil de manejar
x_train = x_train / 255.0
x_test = x_test / 255.0

# Aplanar las imágenes (de 28x28 a 784)
x_train = x_train.reshape(-1, 28 * 28) # Pasamos a un vector de 784 pixeles
x_test = x_test.reshape(-1, 28 * 28)  # Lo mismo con los datos de text

# Crear el modelo
model = models.Sequential([
    layers.Dense(128, activation='relu', input_shape=(784,)), # Primera capa, 128 neuronas y activación Relu
    layers.Dropout(0.2), # Apagamos un 20% neuronas para evitar overfitting (sobreentenamiento)
    layers.Dense(10, activation='softmax') # Capa de salida usando activación softmax
])

# Compilar el modelo, usand el optimizador adam (ajusta learning rate automáticamente)
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Entrenar el modelo con 5 épocas (recorre 5 veces el dataset mnist)
model.fit(x_train, y_train, epochs=5)

# Evaluar el modelo
test_loss, test_acc = model.evaluate(x_test, y_test) # Le pasamos los datos de text para evaluarlo
print(f'Test accuracy: {test_acc:.4f}') # Debería ser en torno a un 97-98% de accuracy
```
